## Segundo passo: Importar biblioteca
Requests é uma das bibliotecas que usamos para fazer requisições HTTP (GET, POST, etc..)

```python
import requests
import pandas as pd
```

## Terceiro passo: Fazer uma requisição para a PokéAPI

`url`: endpoint da PokéAPI que retorna informações sobre o Pikachu.

`requests.get(url)`: envia uma requisição HTTP do tipo GET.

`resposta`: objeto que contém o resultado da requisição.

```python
url = "https://servicodados.ibge.gov.br/api/v1/localidades/municipios"
resposta = requests.get(url)
```

## Quarto passo: Verificar se a requisição deu certo

O código de status 200 significa que a requisição foi bem-sucedida.

Se o código for 404, significa que o recurso não foi encontrado.

```python
print(resposta.status_code)
```

## Quinto passo: Converter resposta para JSON

`json()` transforma a resposta em um dicionário Python.

Agora podemos acessar os campos como se fossem chaves de um dicionário.

```python
dados = resposta.json()
```

# Sexto passo: Explorar os dados

Saída esperada:

Nome: pikachu

Altura: 4

Peso: 60

Tipos: ['electric']

```python
df = pd.DataFrame(dados)
print(df.tail())
```

```python
print("Nome:", dados["name"])
print("Altura:", dados["height"])
print("Peso:", dados["weight"])
print("Tipos:", [t["type"]["name"] for t in dados["types"]])
```

## Função para buscar qualquer Pokémon

```python
def get_pokemon(name):
    url = f"https://pokeapi.co/api/v2/pokemon/{name.lower()}"
    resposta = requests.get(url)
    if resposta.status_code == 200:
        dados = resposta.json()
        return {
            "Nome": dados["name"],
            "Altura": dados["height"],
            "Peso": dados["weight"],
            "Tipos": [t["type"]["name"] for t in dados["types"]]
        }
    else:
        return {"error": "Pokémon não encontrado"}
```

```python
get_pokemon("bulbasaur")
```
